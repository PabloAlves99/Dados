```python
full_dados_csv = []
with open(path_csv, 'r') as file:
    data_csv = csv.DictReader(file)
    for row in data_csv:
        full_dados_csv.append(row)
```

```python
mapeamento = {
    'Nome do Item': 'Nome do Produto',
    'Classificação do Produto': 'Categoria do Produto',
    'Valor em Reais (R$)': 'Preço do Produto (R$)',
    'Quantidade em Estoque': 'Quantidade em Estoque',
    'Nome da Loja': 'Filial'
}
```

```python
new_data_csv = []

for item in full_dados_csv:
    map_v = {}
    for k, v in item.items():
        if k in mapeamento.keys():
            map_v[mapeamento[k]] = v
        else:
            map_v[k] = v
    new_data_csv.append(map_v)
```

```python
print(full_dados_csv.__len__())
print(new_data_csv.__len__())
print(data_json.__len__())
```

```python
print(new_data_csv[0].keys())
print(new_data_csv[0].keys())
print(data_json[0].keys())
```

```python
combined_list = []
combined_list.extend(data_json)
combined_list.extend(new_data_csv)
```

```python
len(combined_list)
```

```python
nomes_colunas = list(combined_list[-1].keys())
nomes_colunas
```

```python
dados_comninados_tabela = [nomes_colunas]

for row in combined_list:
    linha = []
    
    for col in nomes_colunas:
        linha.append(row.get(col, 'Indisponivel'))
        
    dados_comninados_tabela.append(linha)
```

```python
len(dados_comninados_tabela)
```

```python
path_dados_combinados = '../data_processed/dados_combinados.csv'
with open(path_dados_combinados, 'w') as file:
    writer = csv.writer(file)
    writer.writerows(dados_comninados_tabela)
```
